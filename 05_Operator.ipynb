# OPERATOR
# An operator is a symbol or keyword that tells Python to perform an operation.

```python
# Arithmetic
a = 78.5
b = 2
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)
```

```python
print("Python " * 3)
```

```python
# Comparison
a=10
b=3
print("Equal:", a == b)
print("Not Equal:", a != b)
print("Greater:", a > b)
print("Smaller:", a < b)
print("Greater or Equal:", a >= b)
print("Smaller or Equal:", a <= b)
```

```python
#  ASSIGNMENT OPERATORS
# Used to assign/update values
number = 10
print("Original:", number)
number += 5
print( number)
number -= 3
print( number)
number *= 2
print( number)
number /= 4
print( number)
number //= 2
print( number)
number %= 3
print( number)
number **= 2
print( number)
```

```python
# Logical
a=33
b=22
print("Logical AND:", a > 5 and b < 5)
print("Logical OR:", a < 5 or b < 5)
print("Logical NOT:", not(a > b))
```

```python

# Membership
numbers = [10, 20, 30]
print("10 exists:", 10 in numbers)
print("50 exists:", 50 in numbers)
```

```python

# BITWISE OPERATORS
# Work with numbers at the binary/bit level
p = 5
q = 3
print("AND &:", p & q)
print("OR |:", p | q)
print("XOR ^:", p ^ q)
print("NOT ~:", ~p)
print("Left Shift <<:", p << 1)
print("Right Shift >>:", p >> 1)
```

```python
# IDENTITY OPERATORS
# Check whether two variables refer to the same object
value = None
print("is:", value is None)
print("is not:", value is not None)
```

```python
student_age = 20
student_marks = 85

is_adult = student_age >= 18
is_passed = student_marks >= 40
is_eligible = student_age >= 18 and student_marks >= 60

print("Age:", student_age)
print("Marks:", student_marks)
print("Adult:", is_adult)
print("Passed:", is_passed)
print("Eligible:", is_eligible)
```
