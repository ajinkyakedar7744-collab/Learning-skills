```python
# A set is a collection of unique values.
# Main features:
# 1. Duplicate values are automatically removed.
# 2. Sets are unordered.
# 3. Sets are mutable, so we can add/remove values.
# 4. Sets do NOT use indexing.
```

```python
# CREATING A SET
numbers = {10, 20, 30, 40}
print("Set:", numbers)
print("Type:", type(numbers))
```

```python
# DUPLICATE VALUES
numbers = {10, 20, 10, 30, 20, 40}
print("Set:", numbers)
```

```python
# ADD()
numbers = {10, 20, 30}
# add() adds ONE new value to the set.
numbers.add(40)
print("After add:", numbers)
```

```python
#ADD DUPLICATE VALUE
numbers.add(20)
# 20 already exists.
# Set will NOT create another 20.
print("After duplicate:", numbers)
```

```python
#  UPDATE()
numbers = {10, 20}
# update() adds MULTIPLE values.
numbers.update([30, 40, 50])
print("After update:", numbers)
```

```python
#  REMOVE()
numbers = {10, 20, 30, 40}
# remove() deletes a specific value.
numbers.remove(30)
print("After remove:", numbers)

# If the value does not exist, remove() gives an error.
# numbers.remove(100)    # Error
```

```python
# 9. DISCARD()
numbers = {10, 20, 30}
# discard() also removes a value.
numbers.discard(20)
print("After discard:", numbers)
# Difference:
# remove()  error if value doesn't exist
# discard()  no error if value doesn't exist
numbers.discard(100)    # ✅ No error
```

```python
# POP()
numbers = {10, 20, 30, 40}
# pop() removes and returns an arbitrary element.
# Since a set is unordered, don't expect a specific value.
removed = numbers.pop()
print("Removed:", removed)
print("Set:", numbers)
```

```python
# CLEAR()
numbers = {10, 20, 30}
# clear() removes all elements.
numbers.clear()
print("After clear:", numbers)
```

```python
#  LENGTH OF SET
numbers = {10, 20, 30, 40}
# len() tells how many unique values are present.
print("Length:", len(numbers))
```

```python
#  CHECK WHETHER VALUE EXISTS
numbers = {10, 20, 30, 40}
# 'in' checks whether a value exists.
print(20 in numbers)
print(100 in numbers)
# 'not in' checks whether a value does not exist.
print(100 not in numbers)
```

```python
#  LOOP THROUGH A SET
numbers = {10, 20, 30, 40}
# We can use a for loop to visit every value.
for value in numbers:
    print("Value:", value)
# Remember:
# The order is not guaranteed.
```

```python
#  UNION
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
# Union combines values from BOTH sets.
# Duplicate values appear only once.
result = a | b
print("Union:", result)
# We can also use the union() method.
result = a.union(b)
print("Union:", result)
```

```python
# INTERSECTION
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
# Intersection gives values common to BOTH sets.
result = a & b
print("Intersection:", result)
# Same operation using method:
result = a.intersection(b)
print("Intersection:", result)
```

```python
# DIFFERENCE
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
# a - b means:
# Values present in a but NOT present in b.
result = a - b
print("Difference a-b:", result)
# b - a means:
# Values present in b but NOT present in a.
result = b - a
print("Difference b-a:", result)
```

```python
#  SYMMETRIC DIFFERENCE
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
# Symmetric difference gives values that are
# NOT common to both sets.
result = a ^ b
print("Symmetric difference:", result)
```

```python
# SUBSET
a = {1, 2}
b = {1, 2, 3, 4}
# a is a subset of b because every value of a
# exists inside b.
print(a.issubset(b))
```

```python
# DISJOINT
a = {1, 2, 3}
b = {4, 5, 6}
# Two sets are disjoint when they have
# NO common values.
print(a.isdisjoint(b))
```

```python
# REMOVE DUPLICATES FROM A LIST
numbers = [10, 20, 10, 30, 20, 40, 10]
# Converting a list to a set removes duplicates.
unique_numbers = set(numbers)
print("Unique:", unique_numbers)
```

```python
# FIND COMMON VALUES
students_python = {"Ajinkya", "Ayush", "Piyush", "Harsh"}
students_sql = {"Ajinkya", "Ayush", "Piyush"}
# Students learning BOTH Python and SQL.
common = students_python & students_sql
print("Learning both:", common)
```

```python
# STUDENTS ONLY LEARNING PYTHON
only_python = students_python - students_sql
print("Only Python:", only_python)
```

```python
# ALL STUDENTS
all_students = students_python | students_sql
print("All students:", all_students)
```

```python
# FIND UNIQUE USER INPUT
data = input("Enter numbers separated by spaces: ")
# split() separates the input.
# set() removes duplicate values.
numbers = set(data.split())
print("Unique values:", numbers)
```

```python
#  EMPTY SET
# {} creates an EMPTY DICTIONARY, NOT an empty set.
empty_dict = {}
print(type(empty_dict))
# To create an empty set, use set().
empty_set = set()
print(type(empty_set))
```
