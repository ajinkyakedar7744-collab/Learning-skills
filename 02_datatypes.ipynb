# INTEGER DATATYPE
## An integer is a whole number

```python
age = 23
marks = 95
temperature = -5
students = 100
print(type(age))
print(type(marks))
print(type(temperature))
print(type(students))
```

# float

# A float represents a decimal number.

```python
height = 5.8
weight = 77.5
price = 99.99
a=10.0
print(type(height))
print(type(weight))
print(type(price))
print(type(a))
```

# String — str
# A string is text.

```python
name = "Ajinkya"
language = 'Python'
message = "I am learning Python"
x = "100"
print(type(name))
print(type(language))
print(type(message))
print(type(x))
```

# Boolean — bool

```python
is_student = True
is_teacher = False
print(type(is_student))
print(type(is_teacher))
```

# None — NoneType
# None means no value / absence of a value.

```python
result = None
print(type(result))
```

```python
name = "Ajinkya"
height = 5.8
is_student = True

print(type(name))
print(type(height))
print(type(is_student))
```

# Complex

```python
number = 2 + 3j
print(type(number))
```

# LIST

```python
marks = [85, 90, 78, 92]
print(type(marks))
```

# TUPLES

```python
subjects = ("Python", "SQL", "Maths")
print(type(subjects))
```

# DICTIONARY

```python
student = {
    "name": "Ajinkya",
    "age": 20,
    "course": "Python"
}
print(type(student))
```
