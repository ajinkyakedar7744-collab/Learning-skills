```python
# STRING CREATION
name = "Ajinkya"
city = 'Mehkar'
message = """Hello!
Welcome to Python.
We are learning Strings."""
```

```python
print(name)
print(city)
print(message)
```

```python
# LENGTH
print(len(name))
print(len(city))
print(len(message))
```

```python
# INDEXING
print("First character:", name[0])
print("Second character:", name[1])
print("Last character:", name[-1])
print("Second last character:", name[-2])
```

```python
# SLICING
print("First 5 characters:", name[0:5])
print("Characters 2 to 6:", name[2:7])
print("From beginning to 5:", name[:5])
print("From 6 to end:", name[6:])
print("Every second character:", name[::2])
print("Reverse:", name[::-1])
```

```python
# STRING METHODS
text = "  Hello Python World  "
print("Original:", text)
print("Upper:", text.upper())
print("Lower:", text.lower())
print("Strip:", text.strip())
print("Replace:", text.replace("Python", "Data Science"))
print("Count of 'o':", text.count("o"))
print("Find 'Python':", text.find("Python"))
print("Starts with spaces:", text.startswith("  "))
print("Ends with spaces:", text.endswith("  "))
```

```python
# CHECKING STRING CONTENT
word = "Python"
number = "12345"
print("Is alphabet:", word.isalpha())
print("Is digit:", number.isdigit())
print("Is alphanumeric:", "Python123".isalnum())
```

```python
#  SPLIT
sentence = "Python is easy to learn"
words = sentence.split()
print("Sentence:", sentence)
print("Words:", words)
```

```python
# JOIN
joined = "-".join(words)
print("Joined words:", joined)
```

```python
# STRING FORMATTING
name = "Ajinkya"
age = 22
course = "Data Science"

print(f"My name is {name}.")
print(f"I am {age} years old.")
print(f"I am learning {course}.")
```

```python
#  CONCATENATION
first_name = "Ajinkya"
last_name = "Kedar"

full_name = first_name + " " + last_name

print("Full name:", full_name)
```

```python
# REPETITION
print("Python " * 3)
```

```python
# MEMBERSHIP

language = "Python"

print("'P' in language:", "P" in language)
print("'Java' in language:", "Java" in language)
print("'Python' in language:", "Python" in language)

```

```python
#  STRING COMPARISON

a = "apple"
b = "banana"

print("apple == banana:", a == b)
print("apple != banana:", a != b)
print("apple < banana:", a < b)
```

```python
#  ESCAPE CHARACTERS

print("Hello\nWorld")
print("Python\tProgramming")
print("He said, \"Python is easy.\"")
print("It's a great language.")
```

```python

# STRING IMMUTABILITY
text = "Python"
# We cannot directly change one character:
# text[0] = "J"       # This gives TypeError
# Instead, create a new string
text = "J" + text[1:]
print("New string:", text)
```
