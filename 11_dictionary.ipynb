```python
# A dictionary stores data in KEY : VALUE pairs.
```

```python
#  CREATE A DICTIONARY
student = {
    "name": "Ajinkya",
    "age": 20,
    "course": "Python"
}

print("Student:", student)
print("Type:", type(student))
```

```python
# ACCESS VALUE USING KEY
print("Name:", student["name"])
print("Age:", student["age"])
print("Course:", student["course"])
```

```python
# 4. ADD A NEW KEY-VALUE PAIR
# If the key does not exist, Python creates it.

student["city"] = "Mehkar"

print("After adding city:", student)
```

```python
# UPDATE AN EXISTING VALUE
# If the key already exists, its value is changed.
student["age"] = 22
print("Updated age:", student)
```

```python
# UPDATE MULTIPLE VALUES
# update() can add or modify multiple key-value pairs.

student.update({
    "age": 22,
    "course": "Data Science"
})

print("Updated dictionary:", student)
```

```python
#  get()
# get() is used to safely access a value.
print(student.get("name"))

# If the key doesn't exist, get() returns None
# instead of immediately giving a KeyError.
print(student.get("phone"))

# We can also provide a default value.
print(student.get("phone", "Not available"))
```

```python
# CHECK WHETHER A KEY EXISTS
# 'in' checks KEYS in a dictionary.

if "name" in student:
    print("Name key exists")


if "phone" not in student:
    print("Phone key does not exist")
```

```python
# DELETE USING del

student = {
    "name": "Ajinkya",
    "age": 22,
    "city": "Mehkar"
}

# del removes the specified key-value pair.

del student["city"]

print("After del:", student)
```

```python
#  pop()
student = {
    "name": "Ajinkya",
    "age": 22,
    "city": "Mehkar"
}

# pop() removes the specified key
# and returns its value.

city = student.pop("city")

print("Removed:", city)
print("Dictionary:", student)
```

```python
#  popitem()
student = {
    "name": "Ajinkya",
    "age": 22,
    "course": "Python"
}

# popitem() removes the last inserted key-value pair.

item = student.popitem()

print("Removed item:", item)
print("Dictionary:", student)
```

```python
#  CLEAR()
student = {
    "name": "Ajinkya",
    "age": 22
}

# clear() removes everything.

student.clear()

print("After clear:", student)
```

```python
#  len()
student = {
    "name": "Ajinkya",
    "age": 22,
    "course": "Python"
}

# len() returns the number of key-value pairs.

print("Number of items:", len(student))
```

# DICTIONARY METHODS


```python
# 14. keys()
student = {
    "name": "Ajinkya",
    "age": 22,
    "course": "Python"
}
# keys() returns all keys.
print("Keys:", student.keys())
```

```python
# values()
# values() returns all values.
print("Values:", student.values())
```

```python
#  items()
# items() returns key-value pairs.
print("Items:", student.items())
```

```python
# LOOP THROUGH KEYS
for key in student:
    print("Key:", key)
```

```python
#  LOOP THROUGH VALUES
for value in student.values():
    print("Value:", value)
```

```python
#  LOOP THROUGH KEY AND VALUE
for key, value in student.items():
    print(key, "=", value)
```

```python
# STUDENT MARKS
marks = {
    "Python": 85,
    "SQL": 78,
    "Excel": 90,
    "Statistics": 82
}

print("Python marks:", marks["Python"])
print("SQL marks:", marks["SQL"])
```

```python
# CALCULATE TOTAL MARKS
total = 0
for mark in marks.values():
    total += mark

print("Total:", total)
```

```python
#  CALCULATE AVERAGE
average = total / len(marks)
print("Average:", average)
```

```python
# FIND HIGHEST MARK
highest = 0

for subject, mark in marks.items():

    # Compare every mark with highest.

    if mark > highest:
        highest = mark

print("Highest mark:", highest)
```

```python
#  FIND SUBJECT WITH HIGHEST MARK
highest = 0
highest_subject = ""

for subject, mark in marks.items():

    if mark > highest:
        highest = mark
        highest_subject = subject

print("Highest subject:", highest_subject)
print("Highest mark:", highest)
```

```python
#  FILTER MARKS
# Create a new dictionary containing
# subjects where marks are >= 80.

good_marks = {}

for subject, mark in marks.items():

    if mark >= 80:
        good_marks[subject] = mark

print("Marks >= 80:", good_marks)
```

# NESTED DICTIONARY


```python
# DICTIONARY INSIDE DICTIONARY

students = {
    "student1": {
        "name": "Ajinkya",
        "age": 22,
        "marks": 85
    },

    "student2": {
        "name": "Rahul",
        "age": 21,
        "marks": 90
    }
}

print(students)
```

```python

# ACCESS NESTED DICTIONARY

print(students["student1"]["name"])
print(students["student2"]["marks"])
```

```python
#  LOOP THROUGH NESTED DICTIONARY

for student_id, data in students.items():

    print("ID:", student_id)
    print("Name:", data["name"])
    print("Age:", data["age"])
    print("Marks:", data["marks"])
```

```python
# CREATE DICTIONARY FROM USER INPUT
student = {}

name = input("Enter name: ")
age = int(input("Enter age: "))
course = input("Enter course: ")

# Store the entered information.

student["name"] = name
student["age"] = age
student["course"] = course

print("Student information:", student)
```

```python
# BASIC DICTIONARY COMPREHENSION
numbers = [1, 2, 3, 4, 5]
squares = {
    x: x ** 2
    for x in numbers
}
print("Squares dictionary:", squares)
```

```python
# FILTER USING DICTIONARY COMPREHENSION
numbers = [1, 2, 3, 4, 5, 6]
# Create a dictionary only for even numbers.
even = {
    x: x ** 2
    for x in numbers
    if x % 2 == 0
}
print("Even squares:", even)
```
