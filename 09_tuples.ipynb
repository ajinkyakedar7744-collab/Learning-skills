```python
# A tuple is a collection of multiple values.
# Tuples are ordered, meaning each value has an index.
```

```python
# Main difference:
# List   -> [10, 20, 30] -> Mutable (can change)
# Tuple  -> (10, 20, 30) -> Immutable (cannot change)
```

```python
# 1. CREATING A TUPLE
numbers = (10, 20, 30, 40, 50)
print("Tuple:", numbers)
```

```python
print("Type:", type(numbers))
```

```python
#  TUPLE INDEXING
print("First element:", numbers[0])
print("Last element:", numbers[-1])
print("Element at index 2:", numbers[2])
print("Second last:", numbers[-2])
print("Second:", numbers[1])
print("Last:", numbers[4])

```

```python
#  TUPLE SLICING
print("First three:", numbers[0:3])
print("Middle:", numbers[1:4])
print("From index 2:", numbers[2:])
print("Last two:", numbers[-2:])
print("All except last:", numbers[:-1])
print("All except first:", numbers[1:])
print("All except first and last:", numbers[1:-1])
print("All elements:", numbers[:])
```

```python
# LOOP THROUGH A TUPLE
# for loop can visit every tuple value.
for value in numbers:
    print("Value:", value)
    
```

```python
# LENGTH OF TUPLE
print("Length:", len(numbers))
```

```python
#  CHECK WHETHER VALUE EXISTS
# 'in' checks whether a value exists in the tuple.
print(30 in numbers)
print(100 in numbers)
# 'not in' checks whether a value does NOT exist.
print(100 not in numbers)
```

```python
#  SINGLE-VALUE TUPLE
# A comma is required to create a one-element tuple.
a = (10,)
print(a)
print(type(a))
# Without comma, Python considers it an integer.
b = (10)
print(b)
print(type(b))
```

```python
#  MIXED DATA TYPES
# A tuple can contain different types of data.
student = ("Ajinkya", 20, 85.5, True)
print(student)
```

```python
# TUPLE WITH LIST
# A tuple can contain another collection.
data = ("Ajinkya", [10, 20, 30])
print(data)
```

```python
# CONCATENATION
# Two tuples can be joined using +.
a = (1, 2, 3)
b = (4, 5, 6)
result = a + b
print("Joined tuple:", result)
```

```python
#  REPETITION
# * repeats the tuple multiple times.
numbers = (1, 2)
result = numbers * 3
print("Repeated:", result)
```

```python
# COUNT()
# count() tells how many times a value appears.
numbers = (10, 20, 10, 30, 10, 40)
print("10 appears:", numbers.count(10), "times")
```

```python
#  INDEX()
# index() returns the position of the first occurrence.
numbers = (10, 20, 30, 40)
print("Position of 30:", numbers.index(30))
```

```python
# PACKING
# Putting multiple values into one tuple is called packing.
student = "Ajinkya", 20, "Python"
print(student)
print(type(student))
```

```python
#  UNPACKING
# We can extract tuple values into separate variables.
student = ("Ajinkya", 20, "Python")
name, age, language = student
print("Name:", name)
print("Age:", age)
print("Language:", language)
# Number of variables should normally match
# the number of tuple elements.
```

```python
#  SWAPPING VARIABLES
a = 10
b = 20
# Python can swap values using tuple unpacking.
a, b = b, a
print("a:", a)
print("b:", b)
```

```python
# NESTED TUPLE
# A tuple can contain other tuples.
data = (
    (10, 20),
    (30, 40),
    (50, 60)
)

print(data)
# Access the first inner tuple.
print(data[0])
# Access 20.
print(data[0][1])
```

```python
#  LOOP THROUGH NESTED TUPLE
for row in data:
    # Each row is an inner tuple.
    for value in row:
        print(value)
```

```python
#  CONVERT LIST TO TUPLE

numbers = [10, 20, 30, 40]

# tuple() converts the list into a tuple.

numbers_tuple = tuple(numbers)

print(numbers_tuple)
print(type(numbers_tuple))
```

```python
# CONVERT TUPLE TO LIST
numbers = (10, 20, 30, 40)
# list() converts the tuple into a list.
numbers_list = list(numbers)
print(numbers_list)
print(type(numbers_list))
```

```python
# Tuple has only two main methods:
# 1. count()
# 2. index()
numbers = (10, 20, 10, 30, 10)
print(numbers.count(10))
print(numbers.index(20))
```
