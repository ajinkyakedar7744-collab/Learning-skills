```python
numbers = [30, 10, 20, 40, 10]
print(numbers)
```

```python
# 1. append()
numbers.append(50)
print("append:", numbers)
```

```python
# 2. insert()
numbers.insert(1, 15)
print("insert:", numbers)
```

```python
# 3. extend()
numbers.extend([60, 70])
print("extend:", numbers)
```

```python
# 4. remove()
numbers.remove(10)
print("remove:", numbers)
```

```python
# 5. pop()
removed = numbers.pop()
print("pop removed:", removed)
print("after pop:", numbers)
```

```python
# 6. index()
position = numbers.index(20)
print("index of 20:", position)
```

```python
# 7. count()
frequency = numbers.count(10)
print("count of 10:", frequency)
```

```python
# 8. sort()
numbers.sort()
print("sorted:", numbers)
```

```python
# 9. reverse()
numbers.reverse()
print("reversed:", numbers)
```

```python
# 10. copy()
new_numbers = numbers.copy()
print("copy:", new_numbers)
```

```python
# 11. len()
print("length:", len(numbers))
```

```python
# 12. in
if 20 in numbers:
    print("20 exists")
```

```python

# 13. not in
if 100 not in numbers:
    print("100 does not exist")
```

```python
# 14. clear()
numbers.clear()
print("after clear:", numbers)
```
