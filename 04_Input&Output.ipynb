# SIMPLE CALCULATOR


```python

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
# String Concatenation
# You can join strings using +
first_name = "Ajinkya"
last_name = "Kedar"
full_name = first_name + "" + last_name
full_name_without_space = first_name + last_name
print(full_name)
print(full_name_without_space)
```

```python
# Multiple inputs in ONE line
x, y = map(int, input("Enter two numbers: ").split(","))
print("x =", x)
print("y =", y)
```

```python
# Multiple integer inputs
p, q, r, s, t, u, v = map(int, input("Enter seven numbers: ").split())
print("Numbers:", p, q, r, s, t, u, v)
```

```python
# Multiple FLOAT inputs
num1, num2 = map(float, input("Enter two decimal numbers: ").split())
print("Number 1:", num1)
print("Number 2:", num2)
```

```python
# Multiple STRING inputs
str1, str2 = map(str, input("Enter two strings: ").split())
print("String 1:", str1)
print("String 2:", str2)
```

```python
# Multiple Bool inputs
bool1, bool2 = map(bool, input("Enter two boolean values (True/False): ").split())
print("Boolean 1:", bool1)
print("Boolean 2:", bool2)
```

```python
# F-STRING 
name = input("Enter your name: ")
city = input("Enter your city: ")
course = input("Enter your course: ")
print(f"My name is {name}, I live in {city}, and I am learning {course}.")
```

```python
# Calculate Average
mark1 = float(input("Enter mark 1: "))
mark2 = float(input("Enter mark 2: "))
mark3 = float(input("Enter mark 3: "))
average = (mark1 + mark2 + mark3) / 3
print(f"Average marks = {average:.2f}")
```

```python
# Calculate Rectangle Area
length = float(input("Enter length: "))
width = float(input("Enter width: "))
area = length * width
print(f"Rectangle area = {area}")
```

```python
# Temperature Conversion
# Celsius → Fahrenheit
celsius = float(input("Enter temperature in Celsius: "))
fahrenheit = (celsius * 9 / 5) + 32
print(f"Temperature = {fahrenheit:.2f}°F")
```

```python
# BMI Calculation
weight = float(input("Enter weight in kg: "))
height = float(input("Enter height in meters: "))
bmi = weight / (height ** 2)
print(f"Your BMI is {bmi:.2f}")
```

```python
# Output formatting
price = 99.56789
print(f"Price: ₹{price:.2f}")
# .2f → 2 digits after decimal
```

```python
#  Check the type of input
value =input("Enter anything: ")
print("Value:", value)
print("Data type:", type(value))
# input() ALWAYS returns str
```
