```python
# A prime number has exactly two factors
n = int(input("Enter number: "))
count = 0
for i in range(1, n + 1):
    if n % i == 0:
        count += 1
if count == 2:
    print("Prime")
else:
    print("Not Prime")
```

```python
# Finds the largest among entered numbers
n = int(input("How many numbers? "))
maximum = None
for i in range(n):
    number = int(input("Enter number: "))
    if maximum is None or number > maximum:
        maximum = number
print("Largest =", maximum)
```

```python
# Finds the smallest among entered numbers
n = int(input("How many numbers? "))
minimum = None
for i in range(n):
    number = int(input("Enter number: "))
    if minimum is None or number < minimum:
        minimum = number
print("Smallest =", minimum)
```

```python
#for loop with break
# Stops the loop when i becomes 5
for i in range(1, 11):
    if i == 5:
        break
    print(i)
```

```python
# for loop with continue
# Skips number 5
for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```

```python
#Nested for loop
# Loop inside another loop
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
```

```python
#Nested loop with pattern
# Prints a triangle of stars
n = int(input("Enter rows: "))
for i in range(1, n + 1):
    print("*" * i)
```

```python
# Square pattern
# Prints a square of stars
n = int(input("Enter rows: "))
for i in range(n):
    print("*" * n)
```

```python
# number pattern
# Prints increasing numbers
n = int(input("Enter rows: "))
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

```python
# Reverse number pattern
# Prints decreasing numbers
n = int(input("Enter rows: "))
for i in range(n, 0, -1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

```python
# Multiplication tables from 1 to n
# Prints multiple multiplication tables

n = int(input("Enter n: "))

for i in range(1, n + 1):
    print("Table of", i)

    for j in range(1, 11):
        print(i, "x", j, "=", i * j)
```

```python
#for loop with else
# else executes when the loop finishes normally
for i in range(1, 6):
    print(i)
else:
    print("Loop completed")
```

```python
#for loop with break and else
# else does not execute when break occurs
for i in range(1, 6):
    if i == 3:
        break
    print(i)
else:
    print("Loop completed")
```
